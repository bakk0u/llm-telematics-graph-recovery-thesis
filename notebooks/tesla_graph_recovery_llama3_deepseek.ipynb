### Prompt Decomposition for Telematics Graph Recovery (Ollama)
##### v3.7.1 — Tesla Use Case for Deepseek-r1:8b and llama3.1:8b

**Goal of the code.** Run a controlled comparison of *17 prompting
strategies* × *2 local LLMs* (`deepseek-r1:8b`, `llama3.1:8b`) on the
**Tesla Full EV** telematics dataset, score each result against a
peer-reviewed-style Causal Discovery ground truth, and emit a single
*Evaluation Matrix of Causal Discovery* that ranks every
`(model, strategy)` cell on 8 standard metrics.

> **What changed vs v3.7** — Mercedes-Benz Hybrid use case removed; model
> roster reduced from 10 to 2 (`deepseek-r1:8b`, `llama3.1:8b`). All other
> machinery (parser, 8-metric evaluator, repeated-run benchmark, cloud-I/O
> resilience, watchdog) is unchanged from v3.7 — the run simply executes a
> single dataset × 2 models × 17 strategies × N_RUNS instead of the full
> 2 × 10 × 17 × N_RUNS grid.

**Step-by-step workflow.**

1. Configure models (2), datasets (Tesla only), Ollama settings.
2. Define the Ground Truth DAG (with literature citations) for the Tesla
   Use Case and save it as a `.npy` adjacency matrix —
   matching the file convention of `Adjacency_Matrix_Comparison.ipynb`.
3. Define the 17 prompting strategies as data (one entry per strategy).
4. Run every `(dataset, model, strategy)` triple through Ollama; persist each
   result as a `.npy` adjacency matrix and a sidecar `.json`.
5. Build the **Evaluation Matrix**: for every saved `.npy`, compute the
   same 8 metrics used in your existing `Adjacency_Matrix_Comparison.ipynb`
   (`Precision, Recall, F1, SHD, FDR, Pearson, Hamming, Frobenius`).
6. Visualise: full evaluation heatmap, per-LLM heatmap, cross-model F1.
7. Export a knowledge-graph-ready edge list and a JSON-LD bundle.

**Hardware target.** MacBook Pro M1 Max, 32 GB unified memory, macOS, Ollama.

**Reproducibility note.** All randomness in the evaluation pipeline is fully
deterministic given the same Ollama responses; the only stochastic piece is
the LLM itself, controlled via `temperature=0.2`.

## 3. Ground Truth Definition (peer-reviewed)

The Ground Truth DAG is taken verbatim from the user-provided notebook
`Ground_Truth_UC_EV_Energy_Consumption.ipynb`. It encodes 10 directed causal
relationships with sign-and-strength labels and four peer-reviewed citations.

**Citations.**

1. *Battery Voltage and SoC* — DOI: [10.1016/j.jclepro.2022.134381](https://doi.org/10.1016/j.jclepro.2022.134381)
2. *SoC and Energy Consumption / driving style* — DOI: [10.1016/j.trd.2023.103857](https://doi.org/10.1016/j.trd.2023.103857)
3. *Energy Consumption and Electric Remaining* — KIT publication: [1000065370/19170488](https://publikationen.bibliothek.kit.edu/1000065370/19170488)
4. *Mileage and Battery Voltage / battery degradation* — DOI: [10.1016/j.commtr.2024.100148](https://doi.org/10.1016/j.commtr.2024.100148)
5. *Speed, Acceleration, and Energy Consumption* — DOI: [10.1016/j.applthermaleng.2024.124013](https://doi.org/10.1016/j.applthermaleng.2024.124013)

For the **Tesla EV use case** the column set differs (no `batteryvoltage`,
no `sochighlevel`; with `distance_m` as an additional feature). The Tesla
Ground Truth is therefore an *analogue* DAG built from the same physical
principles cited above, restricted to the Tesla feature set. It is clearly
labelled as physics-analogous in the notes, not as independently
peer-reviewed.

> **Note (FIX v3.1):** the `vehiclespeed` channel was originally included
> as a secondary speed signal, but it is empirically a near-duplicate of
> `speed_kmh` on this Tesla CSV (Spearman ρ ≈ 1.0). Two collinear features
> create deterministic causal-discovery artifacts — every algorithm trivially
> infers `speed_kmh ↔ vehiclespeed`, and LLMs do the same — which inflates
> Recall without telling us anything about reasoning quality. `vehiclespeed`
> is therefore dropped from both the feature set and the ground-truth DAG
> (3 edges removed: 9 → 6). See the FIX history at the top of the notebook.


`````python
# ─── Ground truth: Tesla EV (physics-based, peer-reviewed analogue) ────────
# v3.7.1: Mercedes-Benz Hybrid ground truth removed — Tesla is the sole
# use case in this run.
GROUND_TRUTH_TESLA = [
    # FIX v3.1: `vehiclespeed` dropped (collinear with `speed_kmh`).
    # Removed 3 edges: (acceleration_kmh2 -> vehiclespeed),
    #                  (vehiclespeed       -> distance_m),
    #                  (vehiclespeed       -> mileage).
    ("acceleration_kmh2",       "speed_kmh",                "Positive (Strong)"),
    ("speed_kmh",               "distance_m",               "Positive (Strong)"),
    ("speed_kmh",               "mileage",                  "Positive (Strong)"),
    ("speed_kmh",               "electricenergyremaining",  "Negative (Moderate)"),
    ("acceleration_kmh2",       "electricenergyremaining",  "Negative (Strong)"),
    ("electricenergyremaining", "electricremaining",        "Positive (Strong)"),
]
DATASETS["tesla_ev"]["ground_truth_edges"]  = [(u, v) for u, v, _ in GROUND_TRUTH_TESLA]
DATASETS["tesla_ev"]["ground_truth_signed"] = GROUND_TRUTH_TESLA

print(f"Tesla ground truth: {len(GROUND_TRUTH_TESLA)} edges (physics-based)")
`````

`````python
# ─── Build, save, and visualise the Ground Truth adjacency matrices ────────
def build_ground_truth_artifacts(dataset_key: str, results_dir: Path = RESULTS_DIR):
    """
    Construct the Ground Truth DiGraph, save the adjacency matrix as .npy
    (file name matches the user's existing Adjacency_Matrix_Comparison.ipynb
    convention), save a labelled JSON sidecar with sign/strength + citations,
    and produce a labelled PNG of the DAG.
    """
    cfg      = DATASETS[dataset_key]
    features = cfg["features"]
    signed   = cfg["ground_truth_signed"]

    G = nx.DiGraph()
    G.add_nodes_from(features)
    G.add_edges_from([(u, v, {"label": s}) for u, v, s in signed])

    adj = nx.adjacency_matrix(G, nodelist=features).todense()
    adj_df = pd.DataFrame(np.asarray(adj), index=features, columns=features)

    # File name follows the user's existing convention:
    npy_path = results_dir / f"Ground_Truth_{cfg['uc_name']}_adjacency_matrix.npy"
    np.save(npy_path, adj_df.values.astype(np.int8))

    json_path = results_dir / f"Ground_Truth_{cfg['uc_name']}_signed.json"
    json_path.write_text(json.dumps({
        "dataset":   dataset_key,
        "uc_name":   cfg["uc_name"],
        "features":  features,
        "edges_signed": [
            {"source": u, "target": v, "label": s} for u, v, s in signed
        ],
        "n_edges":   len(signed),
    }, indent=2, ensure_ascii=False))

    # Visualise
    fig, ax = plt.subplots(figsize=(11, 8))
    pos = nx.spring_layout(G, seed=42, k=1.4)
    nx.draw(
        G, pos, ax=ax,
        with_labels=True,
        node_size=3000, node_color="skyblue",
        font_size=9, font_weight="bold", arrows=True, arrowsize=18,
    )
    edge_labels = {(u, v): d["label"] for u, v, d in G.edges(data=True)}
    nx.draw_networkx_edge_labels(G, pos, edge_labels=edge_labels,
                                 font_color="firebrick", font_size=8, ax=ax)
    ax.set_title(f"Ground Truth Causal Graph \u2014 {cfg['label']} ({cfg['uc_name']})",
                 fontsize=11)
    plt.tight_layout()
    plt.savefig(results_dir / f"Ground_Truth_{cfg['uc_name']}.png",
                dpi=130, bbox_inches="tight")
    plt.show()

    print(f"\n[{dataset_key}] Adjacency Matrix (Ground Truth) \u2014 saved to:")
    print(f"  {npy_path}")
    print(adj_df)
    return adj_df


for ds_key in DATASETS:
    _ = build_ground_truth_artifacts(ds_key)
`````

## 4. Core utility functions

The pipeline is intentionally split into small, single-responsibility
functions so each step can be unit-tested or replaced independently.


`````python
def safe_name(text: str) -> str:
    """Filesystem-safe slug; identical mapping to v1 so artifacts are stable."""
    return re.sub(r"[^A-Za-z0-9]+", "_", str(text)).strip("_")


# ─── v3.7 — bounded generation + read timeout for reasoning models ─────────
# Per-model output token cap. Reasoning-trained models (deepseek-r1, gpt-oss,
# anything with `<think>` traces, anything `:reasoning` / `:think` tagged)
# need a hard ceiling or they emit thought blocks until something else gives
# up first. Non-reasoning models keep the Ollama default (-1 = unlimited).
MODEL_NUM_PREDICT_OVERRIDE = {
    "deepseek-r1:8b":   1024,   # v3.7.1: the model used in this run
    "deepseek-r1:14b":  1024,
    "deepseek-r1:7b":   1024,
    "deepseek-r1:32b":  1024,
    "gpt-oss:20b":      1024,
    "gpt-oss:7b":       1024,
}


def call_ollama(
    prompt: str,
    model: str,
    temperature: float = TEMPERATURE,
    top_p: float = TOP_P,
    num_ctx: int = NUM_CTX,
    timeout: int = REQUEST_TIMEOUT_S,
) -> dict:
    """Call Ollama /api/generate with bounded generation and a read timeout.

    Two changes vs the original:

    1. `num_predict` is capped per-model via MODEL_NUM_PREDICT_OVERRIDE.
       Reasoning models without this cap can generate <think>...</think>
       traces indefinitely; the cap forces EOS at a known token budget.

    2. `timeout` is now passed as a 2-tuple `(connect, read)` to `requests`.
       A scalar timeout only bounds connection establishment on some
       macOS+OpenSSL builds — slow streams evade it. The read timeout fires
       if NO bytes arrive for `timeout` seconds, which actually bounds the
       call.

    NOTE: temperature/top_p MUST live inside `options`, not at top level
    (the original baseline notebook had them at top level, where Ollama
    silently ignored them).
    """
    options = {
        "temperature": temperature,
        "top_p":       top_p,
        "num_ctx":     num_ctx,
    }
    np_cap = MODEL_NUM_PREDICT_OVERRIDE.get(model)
    if np_cap is not None:
        options["num_predict"] = np_cap

    payload = {
        "model":   model,
        "prompt":  prompt,
        "stream":  False,
        "options": options,
    }
    t0 = time.perf_counter()
    try:
        r = requests.post(OLLAMA_URL, json=payload, timeout=(30, timeout))
        r.raise_for_status()
        data = r.json()
        return {
            "ok": True,
            "response":          data.get("response", ""),
            "latency_s":         time.perf_counter() - t0,
            "eval_count":        data.get("eval_count"),
            "prompt_eval_count": data.get("prompt_eval_count"),
            "total_duration_ns": data.get("total_duration"),
            "error":             None,
        }
    except Exception as e:
        return {
            "ok": False,
            "response":          "",
            "latency_s":         time.perf_counter() - t0,
            "eval_count":        None,
            "prompt_eval_count": None,
            "total_duration_ns": None,
            "error":             f"{type(e).__name__}: {e}",
        }
`````

`````python
def dataset_fingerprint(df: pd.DataFrame, features: list[str]) -> str:
    """
    Compact, semantically rich textual summary of the dataset to send to the
    LLM in place of the raw rows. Shape: ~1.7 KB instead of ~500 KB.
    """
    sub = df[features].dropna()
    desc = sub.describe(percentiles=[0.5]).T[
        ["count", "mean", "std", "min", "50%", "max"]
    ].rename(columns={"50%": "median"})
    desc = desc.round(4)

    rho, _ = spearmanr(sub.values)
    if np.ndim(rho) == 0:
        rho = np.array([[1.0]])
    corr = pd.DataFrame(rho, index=features, columns=features).round(3)

    return (
        "Per-feature statistics (normalized 0..1):\n"
        + desc.to_string()
        + "\n\nSpearman rank-correlation matrix:\n"
        + corr.to_string()
    )


def load_dataset(cfg: dict) -> tuple[pd.DataFrame, str]:
    """Load CSV, keep configured features, return (df, fingerprint_text)."""
    df = pd.read_csv(cfg["csv_path"])
    missing = [c for c in cfg["features"] if c not in df.columns]
    if missing:
        raise ValueError(f"Missing columns in {cfg['csv_path']}: {missing}")
    df = df[cfg["features"]].dropna().reset_index(drop=True)
    fp = dataset_fingerprint(df, cfg["features"])
    return df, fp
`````

`````python
# ─── Robust parser for causal edges from arbitrary LLM output ──────────────
ARROW_RE = re.compile(r"\s*(?:->|\u2192|=>|\u21d2)\s*")
JSON_FENCE_RE = re.compile(r"^\s*```(?:json)?\s*|\s*```\s*$", re.IGNORECASE)


def _strip_fences(text: str) -> str:
    return JSON_FENCE_RE.sub("", text.strip())


def _normalize_token(tok: str, allowed: set[str]) -> str | None:
    t = tok.strip().strip("`*\"' .,;:()[]{}")
    if not t:
        return None
    if t in allowed:
        return t
    low = {a.lower(): a for a in allowed}
    if t.lower() in low:
        return low[t.lower()]
    flat = re.sub(r"[^a-z0-9]", "", t.lower())
    flat_map = {re.sub(r"[^a-z0-9]", "", a.lower()): a for a in allowed}
    return flat_map.get(flat)


def parse_causal_edges(text: str, features: list[str]) -> list[tuple[str, str]]:
    """
    Tolerant to: ``` fences, JSON wrappers, bullets, narrative text, mixed
    arrow glyphs, and case variation. Handles A -> B -> C as two edges.
    """
    if not text:
        return []
    allowed = set(features)
    edges: list[tuple[str, str]] = []
    seen: set[tuple[str, str]] = set()

    raw = _strip_fences(text)

    # Try to find a JSON object with a "causal_discovery" / "edges" list
    for blob in re.findall(r"\{[^\{\}]*?\}", raw, flags=re.DOTALL):
        try:
            obj = json.loads(blob)
        except Exception:
            continue
        if isinstance(obj, dict):
            for key in ("causal_discovery", "edges", "relationships"):
                if key in obj and isinstance(obj[key], list):
                    raw += "\n" + "\n".join(map(str, obj[key]))

    for line in raw.splitlines():
        line = line.strip().strip("-*\u2022").strip()
        if not line or ("\u2192" not in line and "->" not in line and "=>" not in line):
            continue
        parts = ARROW_RE.split(line)
        normed = [_normalize_token(p, allowed) for p in parts]
        for a, b in zip(normed, normed[1:]):
            if a and b and a != b and (a, b) not in seen:
                seen.add((a, b))
                edges.append((a, b))
    return edges


def edges_to_adjacency(edges: list[tuple[str, str]],
                       features: list[str]) -> pd.DataFrame:
    """0/1 adjacency DataFrame, indexed by `features` (cause -> effect)."""
    n = len(features)
    idx = {f: i for i, f in enumerate(features)}
    A = np.zeros((n, n), dtype=np.int8)
    for a, b in edges:
        if a in idx and b in idx:
            A[idx[a], idx[b]] = 1
    return pd.DataFrame(A, index=features, columns=features)
`````

`````python
# ─── 8-metric Evaluation function ──────────────────────────────────────────
# Mirrors `Adjacency_Matrix_Comparison.ipynb` exactly.

def compute_evaluation_metrics(
    predicted: np.ndarray,
    ground_truth: np.ndarray,
    threshold: float = 0.5,
) -> dict:
    """
    Compute the 8 evaluation metrics for a predicted adjacency matrix vs.
    the ground-truth adjacency matrix. Both inputs are square 2-D arrays
    of the same shape; `threshold` binarises (default 0.5).

    Returns
    -------
    dict with keys:
        Precision, Recall, F1 Score,
        SHD (Structural Hamming Distance),
        FDR (False Discovery Rate),
        Pearson Correlation,
        Hamming Distance,
        Frobenius Norm
    """
    if predicted.shape != ground_truth.shape:
        # Defensive: trim to common dimension (mirrors the user's existing code)
        d = min(predicted.shape[0], ground_truth.shape[0])
        predicted    = predicted[:d, :d]
        ground_truth = ground_truth[:d, :d]

    gt_bin   = (ground_truth > threshold).astype(int)
    pred_bin = (predicted    > threshold).astype(int)

    gt_flat   = gt_bin.flatten()
    pred_flat = pred_bin.flatten()

    # Sklearn metrics on flattened binary vectors
    precision = precision_score(gt_flat, pred_flat, zero_division=0)
    recall    = recall_score(   gt_flat, pred_flat, zero_division=0)
    f1        = f1_score(       gt_flat, pred_flat, zero_division=0)

    # Structural Hamming Distance — count of differing entries
    shd = int(np.sum(pred_bin != gt_bin))

    # False Discovery Rate — fraction of predicted edges that are wrong
    fp = int(np.sum((pred_bin == 1) & (gt_bin == 0)))
    n_predicted = int(np.sum(pred_bin == 1))
    fdr = (fp / n_predicted) if n_predicted > 0 else 0.0

    # Pearson correlation between the two flattened binary vectors.
    # Undefined when either vector is constant — guard with a NaN \u2192 0 fallback.
    try:
        if np.std(pred_flat) == 0 or np.std(gt_flat) == 0:
            pearson_corr = 0.0
        else:
            pearson_corr, _ = pearsonr(gt_flat, pred_flat)
            if np.isnan(pearson_corr):
                pearson_corr = 0.0
    except Exception:
        pearson_corr = 0.0

    # Hamming distance on the flattened binary vectors (fraction in [0, 1])
    hamming_dist = float(scipy_hamming(gt_flat, pred_flat))

    # Frobenius norm of the difference (uses the binarised difference for
    # consistency with SHD; numerically equal to sqrt(SHD) here).
    frobenius = float(scipy_norm(pred_bin - gt_bin, "fro"))

    return {
        "Precision":           round(float(precision),    4),
        "Recall":              round(float(recall),       4),
        "F1 Score":            round(float(f1),           4),
        "SHD":                 int(shd),
        "FDR":                 round(float(fdr),          4),
        "Pearson Correlation": round(float(pearson_corr), 4),
        "Hamming Distance":    round(float(hamming_dist), 4),
        "Frobenius Norm":      round(float(frobenius),    4),
    }


# ─── Convenience: compute metrics directly from edge lists ────────────────
def compute_evaluation_metrics_from_edges(
    pred_edges: list[tuple[str, str]],
    gt_edges:   list[tuple[str, str]],
    features:   list[str],
) -> dict:
    pred = edges_to_adjacency(pred_edges, features).values
    gt   = edges_to_adjacency(gt_edges,   features).values
    return compute_evaluation_metrics(pred, gt)
`````

`````python
def plot_causal_graph(
    edges: list[tuple[str, str]],
    title: str,
    features: list[str] | None = None,
    out_path: Path | None = None,
    show: bool = False,
):
    G = nx.DiGraph()
    if features:
        G.add_nodes_from(features)
    G.add_edges_from(edges)

    plt.figure(figsize=(10, 7))
    pos = nx.spring_layout(G, seed=42, k=1.2)
    nx.draw(
        G, pos,
        with_labels=True,
        node_size=2400, node_color="#cfe2f3",
        edge_color="#666",
        font_size=8, font_weight="bold",
        arrows=True, arrowsize=14,
    )
    plt.title(title, fontsize=10)
    plt.tight_layout()
    if out_path is not None:
        out_path.parent.mkdir(parents=True, exist_ok=True)
        plt.savefig(out_path, dpi=120, bbox_inches="tight")
    if show:
        plt.show()
    else:
        plt.close()
`````

## 6. Single-run helper

Run *one* `(dataset, model, strategy)` triple end-to-end. This is the unit
the main experiment loop calls.

Each call writes three artifacts:

- `<strategy>.npy`  — predicted adjacency matrix (this is what the
  Evaluation step in §8 reads)
- `<strategy>.json` — sidecar with predicted edges, scores, raw response,
  latency, error info
- `<strategy>.png`  — predicted-graph plot (optional)


`````python
def run_single(
    dataset_key: str,
    dataset_cfg: dict,
    df: pd.DataFrame,
    fp_text: str,
    model_cfg: dict,
    strategy: dict,
    save_plot: bool = True,
) -> dict:
    """End-to-end execution of one (dataset, model, strategy) cell."""
    features  = dataset_cfg["features"]
    gt_edges  = dataset_cfg["ground_truth_edges"]
    base      = make_base_prompt(
                   dataset_cfg["label"], features, fp_text,
                   dataset_cfg["feature_descriptions"],
                )
    prompt    = strategy["template"](base)
    n_runs    = int(strategy.get("self_consistency_runs", 1))

    raw_responses, latencies, errors = [], [], []
    for _ in range(n_runs):
        out = call_ollama(prompt, model=model_cfg["name"])
        raw_responses.append(out["response"])
        latencies.append(out["latency_s"])
        if not out["ok"]:
            errors.append(out["error"])

    # Self-Consistency: vote on edge-set; otherwise take the single response.
    if n_runs > 1:
        edge_sets   = [tuple(sorted(parse_causal_edges(r, features))) for r in raw_responses]
        most_common = Counter(edge_sets).most_common(1)[0][0]
        edges       = list(most_common)
        used_text   = "\n---\n".join(raw_responses)
    else:
        edges     = parse_causal_edges(raw_responses[0], features)
        used_text = raw_responses[0]

    adj    = edges_to_adjacency(edges, features)
    scores = compute_evaluation_metrics_from_edges(edges, gt_edges, features)

    # ── persist artifacts ──────────────────────────────────────────────────
    model_safe    = safe_name(model_cfg["name"])
    strategy_safe = safe_name(strategy["name"])
    out_dir       = RESULTS_DIR / dataset_key / model_safe
    out_dir.mkdir(parents=True, exist_ok=True)

    np.save(out_dir / f"{strategy_safe}.npy", adj.values.astype(np.int8))
    (out_dir / f"{strategy_safe}.json").write_text(json.dumps({
        "dataset":         dataset_key,
        "model":           model_cfg["name"],
        "strategy":        strategy["name"],
        "kind":            strategy["kind"],
        "features":        features,
        "edges":           edges,
        "scores":          scores,
        "latency_s_total": round(sum(latencies), 3),
        "n_runs":          n_runs,
        "errors":          errors,
        "raw":             used_text,
    }, indent=2, ensure_ascii=False))

    if save_plot:
        plot_causal_graph(
            edges,
            title=f"{model_cfg['name']}  |  {strategy['name']}  |  {dataset_cfg['label']}",
            features=features,
            out_path=out_dir / f"{strategy_safe}.png",
            show=False,
        )

    return {
        "run_id":          uuid.uuid4().hex[:10],
        "dataset":         dataset_key,
        "dataset_label":   dataset_cfg["label"],
        "model":           model_cfg["name"],
        "model_family":    model_cfg["family"],
        "model_size_gb":   model_cfg["size_gb"],
        "strategy":        strategy["name"],
        "strategy_kind":   strategy["kind"],
        "n_runs":          n_runs,
        "latency_s_total": round(sum(latencies), 3),
        "latency_s_mean":  round(sum(latencies) / len(latencies), 3),
        "n_errors":        len(errors),
        "n_pred_edges":    int(adj.values.sum()),
        "edges":           edges,
        "npy_path":        str(out_dir / f"{strategy_safe}.npy"),
        **scores,
    }
`````

## 7. Main experiment loop

This is the only place that actually calls Ollama. The loop is **resumable**:
existing per-cell `*.json` artifacts are skipped on re-run, so you can stop
and continue.

For first-time use, edit `MODELS_TO_RUN` and `DATASETS_TO_RUN` to start with a

## 8. Evaluation Matrix of Causal Discovery

This step does exactly what the user-provided
`Adjacency_Matrix_Comparison.ipynb` did, generalised to all
`(model, strategy)` cells.

For every saved predicted-adjacency `.npy` we:

1. Load the predicted adjacency matrix.
2. Load the dataset-specific Ground Truth adjacency matrix.
3. Compute the same 8 metrics:
   *Precision, Recall, F1 Score, SHD, FDR, Pearson Correlation,
   Hamming Distance, Frobenius Norm*.
4. Stack all rows into a single Evaluation Matrix DataFrame, indexed by
   `<model>_<strategy>`.

This DataFrame **is** the *Evaluation Matrix of Causal Discovery* requested.

A second, identical-content but more analytically convenient long-format
table is also produced (one row per cell, columns include `dataset`, `model`,
`strategy`, plus the 8 metrics).


`````python
EVAL_METRICS_ORDER = [
    "Precision",
    "Recall",
    "F1 Score",
    "SHD",
    "FDR",
    "Pearson Correlation",
    "Hamming Distance",
    "Frobenius Norm",
]


def build_evaluation_matrix(dataset_key: str) -> pd.DataFrame:
    """
    Mirror of `Adjacency_Matrix_Comparison.ipynb`, applied to every
    (model, strategy) cell saved on disk for `dataset_key`.

    The returned DataFrame is the user-facing 'Evaluation Matrix':
        index  : "<model>_<strategy>"   (one row per cell)
        columns: the 8 metrics in EVAL_METRICS_ORDER

    A long-format companion frame (rows include explicit dataset/model/
    strategy/family) is also written to disk for easy slicing.
    """
    cfg     = DATASETS[dataset_key]
    gt_path = RESULTS_DIR / f"Ground_Truth_{cfg['uc_name']}_adjacency_matrix.npy"
    if not gt_path.exists():
        raise FileNotFoundError(f"Ground truth not found at {gt_path}")
    ground_truth = np.load(gt_path)

    rows = []
    for m_cfg in MODELS:
        m_safe = safe_name(m_cfg["name"])
        m_dir  = RESULTS_DIR / dataset_key / m_safe
        if not m_dir.is_dir():
            continue
        for strat in STRATEGIES:
            s_safe = safe_name(strat["name"])
            npy    = m_dir / f"{s_safe}.npy"
            if not npy.exists():
                continue
            predicted = np.load(npy)
            scores    = compute_evaluation_metrics(predicted, ground_truth)
            rows.append({
                "row_label":      f"{m_cfg['name']}_{strat['name']}",
                "dataset":        dataset_key,
                "dataset_label":  cfg["label"],
                "model":          m_cfg["name"],
                "model_family":   m_cfg["family"],
                "model_size_gb":  m_cfg["size_gb"],
                "strategy":       strat["name"],
                "strategy_kind":  strat["kind"],
                **scores,
            })

    long_df = pd.DataFrame(rows)
    if long_df.empty:
        return long_df

    # Save the long-format frame
    long_df.to_csv(RESULTS_DIR / f"evaluation_matrix_{dataset_key}_long.csv",
                   index=False)
    try:
        long_df.to_parquet(
            RESULTS_DIR / f"evaluation_matrix_{dataset_key}_long.parquet",
            index=False,
        )
    except Exception:
        pass

    # User-facing Evaluation Matrix: index=row_label, cols=8 metrics
    eval_df = long_df.set_index("row_label")[EVAL_METRICS_ORDER]
    eval_df.to_csv(RESULTS_DIR / f"evaluation_matrix_{dataset_key}.csv")

    return eval_df


# Build for every dataset run
evaluation_matrices: dict[str, pd.DataFrame] = {}
for ds_key in DATASETS_TO_RUN:
    eval_df = build_evaluation_matrix(ds_key)
    evaluation_matrices[ds_key] = eval_df
    print(f"\n=== Evaluation Matrix \u2014 {DATASETS[ds_key]['label']} ===")
    print(f"Shape: {eval_df.shape}")
    print(eval_df.head(10).to_string())
    print(f"\nSaved to: {RESULTS_DIR / f'evaluation_matrix_{ds_key}.csv'}")
`````

## 9. Visualisations

Three views are produced per dataset, mirroring the seaborn heatmap style
of `Adjacency_Matrix_Comparison.ipynb`:

  cell; columns: the 8 metrics. This is the headline figure.
- **9b — Per-LLM Evaluation Matrix.** One small heatmap per model, rows are

`````python
LOWER_IS_BETTER = {"SHD", "FDR", "Hamming Distance", "Frobenius Norm"}


def _cmap_for(metric: str):
    return "RdYlGn_r" if metric in LOWER_IS_BETTER else "RdYlGn"


def plot_full_evaluation_matrix(eval_df: pd.DataFrame, dataset_key: str):
    """9a \u2014 Full Evaluation Matrix heatmap."""
    if eval_df.empty:
        print(f"[{dataset_key}] no data")
        return
    title = f"Evaluation Matrix of Causal Discovery \u2014 {DATASETS[dataset_key]['label']}"

    # Each metric has a different scale; normalise per-column (z-score \u2192 [0,1])
    # only for the colour mapping. The annotated numbers stay in original units.
    fig, axes = plt.subplots(
        1, len(EVAL_METRICS_ORDER),
        figsize=(2.0 * len(EVAL_METRICS_ORDER), 0.34 * max(len(eval_df), 6) + 2),
        sharey=True,
    )
    if len(EVAL_METRICS_ORDER) == 1:
        axes = [axes]

    for ax, metric in zip(axes, EVAL_METRICS_ORDER):
        col = eval_df[[metric]].copy()
        sns.heatmap(
            col,
            ax=ax,
            annot=True, fmt=".2f" if col[metric].dtype != int else "d",
            cmap=_cmap_for(metric),
            cbar=False, linewidths=0.4, linecolor="white",
            annot_kws={"fontsize": 7},
        )
        ax.set_title(metric, fontsize=9)
        ax.set_xlabel("")
        ax.set_xticks([])
        if ax is axes[0]:
            ax.set_ylabel("model_strategy", fontsize=9)
            ax.tick_params(axis="y", labelsize=7)
        else:
            ax.set_ylabel("")

    plt.suptitle(title, fontsize=12, y=1.005)
    plt.tight_layout()
    out = RESULTS_DIR / f"evaluation_matrix_{dataset_key}_heatmap.png"
    plt.savefig(out, dpi=130, bbox_inches="tight")
    plt.show()
    print(f"Saved: {out}")


for ds_key, eval_df in evaluation_matrices.items():
    plot_full_evaluation_matrix(eval_df, ds_key)
`````

## 10. Knowledge Graph export

Two artifacts:

1. **Long-format edge CSV** `edges_kg.csv`: every predicted edge from every
   `(dataset, model, strategy)` cell, plus a Boolean `is_ground_truth` column.
   This is the form to load into Neo4j / RDF / NetworkX for downstream KG
   analysis.
2. **Per-dataset consensus DAG**: an edge is kept if at least
   `CONSENSUS_THRESHOLD` of the (model, strategy) cells voted for it. This
   is the data-driven *empirical* ground truth that complements the
   peer-reviewed Ground Truth from §3.


`````python
# 10a. Long-format edge CSV ─────────────────────────────────────────────────
edge_rows = []
for r in all_rows:
    gt = set(DATASETS[r["dataset"]]["ground_truth_edges"])
    for (a, b) in r["edges"]:
        edge_rows.append({
            "run_id":          r["run_id"],
            "dataset":         r["dataset"],
            "dataset_label":   r["dataset_label"],
            "model":           r["model"],
            "model_family":    r["model_family"],
            "strategy":        r["strategy"],
            "strategy_kind":   r["strategy_kind"],
            "source":          a,
            "target":          b,
            "is_ground_truth": (a, b) in gt,
        })

edges_kg = pd.DataFrame(edge_rows)
edges_kg.to_csv(RESULTS_DIR / "edges_kg.csv", index=False)
print(f"KG edges saved -> {RESULTS_DIR / 'edges_kg.csv'}  ({len(edges_kg)} rows)")
edges_kg.head()
`````

`````python
# 10b. Consensus DAG per dataset ────────────────────────────────────────────
CONSENSUS_THRESHOLD = 0.50   # keep edges voted for by >= 50% of cells

for ds_key in DATASETS_TO_RUN:
    sub = edges_kg[edges_kg["dataset"] == ds_key] if not edges_kg.empty else edges_kg
    if sub.empty:
        continue
    n_cells = len([r for r in all_rows if r["dataset"] == ds_key])
    if n_cells == 0:
        continue

    counts = (sub.groupby(["source", "target"]).size()
                  .rename("votes").reset_index())
    counts["support"] = counts["votes"] / n_cells
    consensus = counts[counts["support"] >= CONSENSUS_THRESHOLD]                     .sort_values("support", ascending=False)

    out_csv = RESULTS_DIR / f"consensus_{ds_key}.csv"
    consensus.to_csv(out_csv, index=False)

    gt = set(DATASETS[ds_key]["ground_truth_edges"])
    pred = set(zip(consensus["source"], consensus["target"]))
    if gt:
        pred_adj = edges_to_adjacency(list(pred), DATASETS[ds_key]["features"]).values
        gt_adj   = edges_to_adjacency(list(gt),   DATASETS[ds_key]["features"]).values
        s = compute_evaluation_metrics(pred_adj, gt_adj)
        print(f"\n=== Consensus DAG \u2014 {DATASETS[ds_key]['label']} "
              f"(threshold={CONSENSUS_THRESHOLD}, n_cells={n_cells}) ===")
        print(consensus.to_string(index=False))
        print(f"\nvs peer-reviewed Ground Truth:")
        for k, v in s.items():
            print(f"  {k:20s} {v}")
`````

## 11. Repeated-Run Benchmark (N_RUNS)

> **New section — built on top of the single-run pipeline above.**
> The cells in §1–§10 score every `(model, strategy)` cell from a *single*
> Ollama call. The cells below repeat each cell `N_RUNS` times and aggregate
> the results into an averaged benchmark matrix. They write to a separate
> directory (`results/repeated_runs/`) so they do **not** overwrite any of
> the §7 single-run artifacts.

### Why repeat?

LLM output for causal-discovery prompts is *empirically unstable*. The same
prompt, the same model, and the same `temperature=0.2` can produce
different edge sets across runs — a relationship is sometimes detected,
sometimes returned as `0` / "no relationship", sometimes replaced by a
spurious edge. Three sources of variability are at work:

1. **Stochastic decoding.** Even at low temperature the next-token sampler
   is non-deterministic unless `temperature=0` *and* the backend supports
   greedy decoding. Ollama's KV-cache quantisation (`q8_0`) and GPU op
   ordering further prevent bit-exact reproducibility.
2. **Prompt sensitivity.** Each of the 17+3 prompting strategies primes a
   different reasoning path; the run-to-run variance of any one strategy
   is itself a signal worth measuring.
3. **Model response variability.** Smaller models (7B–14B) show higher
   variance than larger ones, but no model in the 7B–27B range is
   completely stable on causal-discovery prompts.

A single LLM run is therefore *not a sufficient observation* for an
academic benchmark. The standard remedy in the LLM-causality literature
(Kıcıman & Sharma 2023; Long et al. 2023; Vashishtha et al. 2024) is to
run each prompt 5–10 times and report the **mean ± std** of every metric,
plus the **edge-detection probability** per causal pair (fraction of runs
in which the LLM detected the edge).

### Why 10 runs (not 5)?

For a Bernoulli edge-detection rate at *p = 0.5*, the standard error is
`SE = √(p(1−p)/n) = 0.5/√n`:

| N_RUNS | SE      |  ddof  |
|-------:|--------:|-------:|
|   5    | 0.2236  |   4    |
|   10   | 0.1581  |   9    |
|   20   | 0.1118  |  19    |

Going from 5 → 10 cuts the standard error by ~30% and pushes the std-estimate
degrees-of-freedom from 4 to 9 — the conventional threshold below which
Student-t tails are noticeably heavy. **10 runs is the defensible round
number for thesis/paper benchmarks.**

### Reproducibility caveat

`np.random.seed(...)` makes the **call sequence** reproducible (which
dataset / model / strategy is dispatched when), but **does not make the LLM
output reproducible**. Ollama backends remain non-deterministic by default.
This is precisely why we average across runs rather than relying on
`temperature=0` alone.

### I/O resilience

This rewrite includes the v3.1–v3.5 fixes for cloud-volume churn, kernel
CWD breakage, atomic file writes, and `np.save`'s `.npy` auto-suffix.
Concretely:

| Failure mode (real, observed)                                  | Fix in this rewrite                                     |
|----------------------------------------------------------------|---------------------------------------------------------|
| `mkdir(parents=True)` recurses through `..` and `.`            | All paths resolved to absolute via `os.path.abspath`    |
| Kernel CWD inode lost to OneDrive sync; `os.getcwd()` raises   | Detect & `chdir` to `Path.home()` before any FS call    |
| `TimeoutError [Errno 60]` from cloud-synced output volume      | Reroute to local `~/causal_discovery_runs/<project>/…`  |
| `pd.to_csv` fails on cloud volume with bogus kwarg error       | Retry with backoff, fall back to pickle                 |
| `np.save("foo.npy.tmp", …)` writes `foo.npy.tmp.npy`           | Temp filename already ends in `.npy`                    |
| One bad call kills the whole `(model, strategy)` cell          | `try/except` per **run**, not per cell                  |

### Artifacts produced

| Artifact                                         | Path                                                              |
|--------------------------------------------------|-------------------------------------------------------------------|
| Per-run adjacency `.npy` (one per repetition)    | `<repeated_runs>/<dataset>/<model>/<strategy>__runNN.npy`         |
| Per-run JSON sidecar (raw response, scores)      | `<repeated_runs>/<dataset>/<model>/<strategy>__runNN.json`        |
| Per-cell edge-probability matrix                 | `<repeated_runs>/<dataset>/<model>/<strategy>__edge_prob.npy`     |
| Raw run-level table                              | `<repeated_runs>/raw_runs.csv`                                    |
| Aggregated table (one row per cell)              | `<repeated_runs>/aggregated_results.csv`                          |
| Average benchmark matrix per dataset             | `<repeated_runs>/average_benchmark_matrix_<dataset>.csv`          |
| Std benchmark matrix per dataset                 | `<repeated_runs>/std_benchmark_matrix_<dataset>.csv`              |
| Detailed statistics                              | `<repeated_runs>/detailed_statistics.csv`                         |
| Heatmaps                                         | `<repeated_runs>/*.png`                                           |

`````python
# ─── §11.1 — Repeated-run configuration & I/O setup ─────────────────────────
# Edit `N_RUNS` here; everything downstream picks it up automatically.

import os as _os
import errno as _errno
import time as _time
import random as _py_random

N_RUNS              = 10            # 10 = recommended; 5 = fast fallback
RANDOM_SEED         = 42            # Python-side reproducibility only
SKIP_EXISTING_RUNS  = False          # resume by skipping per-run *.json that exist
SAVE_PER_RUN_PLOTS  = False         # set True to save thousands of PNGs

np.random.seed(RANDOM_SEED)
_py_random.seed(RANDOM_SEED)


# ── Step 1: heal a broken CWD before any filesystem operation ──────────────
# Jupyter kernels started inside OneDrive/iCloud-synced folders can lose
# their CWD inode mid-session when the cloud provider unmounts/replaces the
# directory (sync cycles, sleep/wake, network blips). Once os.getcwd() fails,
# every relative-path operation in the kernel breaks until we chdir somewhere
# healthy. Path.home() reads $HOME from the environment without touching disk.
def _cwd_is_healthy() -> tuple[bool, str]:
    try:
        cwd = _os.getcwd()
    except (FileNotFoundError, OSError) as e:
        return False, f"{type(e).__name__}: {e}"
    if not _os.path.isdir(cwd):
        return False, f"CWD does not exist on disk: {cwd!r}"
    return True, cwd

_healthy, _cwd_status = _cwd_is_healthy()
if not _healthy:
    _os.chdir(str(Path.home()))
    print(f"   CWD was broken ({_cwd_status}).")
    print(f"   Recovered by chdir to HOME: {_os.getcwd()}")
else:
    print(f"CWD healthy: {_cwd_status}")


# ── Step 2: detect cloud-synced configured path ────────────────────────────
def _is_cloud_synced(p: Path) -> bool:
    s = str(p)
    if any(m in s for m in ("OneDrive", "iCloud", "Dropbox", "Google Drive", "Box Sync")):
        return True
    if s.startswith("/Volumes/") and not s.startswith("/Volumes/Macintosh HD"):
        return True
    return False


# Lexical absolute-path resolver — does NOT stat the filesystem (so it can't
# fail when a cloud-mounted parent component is temporarily unavailable).
def _abspath_no_stat(p: Path) -> Path:
    s = str(p)
    if _os.path.isabs(s):
        return Path(_os.path.normpath(s))
    return Path(_os.path.normpath(_os.path.join(_os.getcwd(), s)))


_configured = _abspath_no_stat(RESULTS_DIR / "repeated_runs")

if _is_cloud_synced(_configured):
    # Use a project-tagged subfolder of HOME so multiple projects don't collide.
    _project_tag = safe_name(Path(str(RESULTS_DIR)).parent.name or "project")
    REPEATED_RUNS_DIR = Path.home() / "causal_discovery_runs" / _project_tag / "repeated_runs"
    SYNC_TARGET_DIR   = _configured     # we'll rsync here once, at the end
    print("   RESULTS_DIR is on a cloud-synced volume — rerouting writes to local disk.")
    print(f"   Configured (cloud) : {_configured}")
    print(f"   Working (local)    : {REPEATED_RUNS_DIR}")
    print(f"   Will sync to cloud after the run completes (see §11.3 epilogue).")
else:
    REPEATED_RUNS_DIR = _configured
    SYNC_TARGET_DIR   = None


# ── Step 3: create the local cache, with a clear error if we can't ─────────
try:
    REPEATED_RUNS_DIR.mkdir(parents=True, exist_ok=True)
except OSError as _e:
    raise OSError(
        f"Could not create REPEATED_RUNS_DIR.\n"
        f"  Target            : {REPEATED_RUNS_DIR}\n"
        f"  Underlying error  : {type(_e).__name__}: {_e}\n"
        f"Hints:\n"
        f"  • errno ENOSPC ({_errno.ENOSPC}) → free disk space\n"
        f"  • errno EACCES ({_errno.EACCES}) → check write permission on {Path.home()}\n"
        f"  • errno EROFS  ({_errno.EROFS})  → target volume is read-only"
    ) from _e


# ── Step 4: chdir into working dir so every relative path downstream
# (matplotlib, pandas, etc.) anchors on a healthy local directory.
_os.chdir(REPEATED_RUNS_DIR)


# ── Step 5: report. os.getcwd() is now safe; wrap it anyway. ───────────────
print(f"\nN_RUNS              = {N_RUNS}")
print(f"RANDOM_SEED         = {RANDOM_SEED}")
print(f"SKIP_EXISTING_RUNS  = {SKIP_EXISTING_RUNS}")
print(f"REPEATED_RUNS_DIR   = {REPEATED_RUNS_DIR}")
try:
    print(f"Working directory   = {_os.getcwd()}")
except OSError as _e:
    print(f"Working directory   = <unavailable: {type(_e).__name__}: {_e}>")
print(f"\nProjected total Ollama calls: "
      f"{len(DATASETS_TO_RUN)} datasets × {len(MODELS_TO_RUN)} models × "
      f"{len(STRATEGIES)} strategies × {N_RUNS} runs "
      f"= {len(DATASETS_TO_RUN) * len(MODELS_TO_RUN) * len(STRATEGIES) * N_RUNS:,} calls")
print("(Strategy_5 Self-Consistency does 5 additional internal calls per outer run.)")
`````

`````python
# ─── §11.2 — Modular functions ──────────────────────────────────────────────
# Public API:
#   _safe_write_json / _safe_save_npy   — atomic, retry-on-transient-error
#   run_single_experiment(...)          — one (dataset, model, strategy) repetition
#   run_repeated_experiments(...)       — N_RUNS reps with watchdog + per-run try/except
#   aggregate_results(...)              — mean/std/edge_prob/n_success/n_zero/...

# v3.7: per-cell wall-clock budget. A reasoning-model hang on one strategy
# cannot now burn more than CELL_BUDGET_S seconds of real time. Remaining
# reps are recorded as failed; SKIP_EXISTING_RUNS picks them up next run.
CELL_BUDGET_S = 15 * 60     # 15 minutes per (model, strategy) cell


# ─────────────────────────────────────────────────────────────────────────────
# Atomic write helpers
# ─────────────────────────────────────────────────────────────────────────────

def _safe_write_json(path: Path, payload: dict,
                     retries: int = 3, backoff: float = 1.5) -> None:
    """Atomic JSON write with retry on ETIMEDOUT/EAGAIN/EBUSY.

    Writes to a sibling ".tmp-<pid>" file then os.replace into place.
    Path.write_text() does not rewrite the path, so the tmp suffix is free.
    """
    tmp = path.with_name(f"{path.name}.tmp-{_os.getpid()}")
    last_err = None
    for k in range(retries):
        try:
            tmp.write_text(json.dumps(payload, indent=2, ensure_ascii=False))
            _os.replace(tmp, path)
            return
        except OSError as e:
            last_err = e
            try:
                if tmp.exists():
                    tmp.unlink()
            except OSError:
                pass
            if e.errno not in (_errno.ETIMEDOUT, _errno.EAGAIN, _errno.EBUSY):
                raise
            _time.sleep(backoff ** k)
    raise last_err


def _safe_save_npy(path: Path, arr: np.ndarray,
                   retries: int = 3, backoff: float = 1.5) -> None:
    """Atomic npy write with retry on ETIMEDOUT/EAGAIN/EBUSY.

    np.save() silently appends '.npy' to any path that doesn't already end
    in '.npy'. So the temp filename MUST already end in '.npy'; otherwise
    np.save writes 'foo.tmp.npy' instead of the requested 'foo.tmp', and
    os.replace fails with FileNotFoundError. We use 'foo.tmp-<pid>.npy'.
    """
    assert path.suffix == ".npy", f"_safe_save_npy expects a .npy path, got {path}"
    tmp = path.with_name(f"{path.stem}.tmp-{_os.getpid()}.npy")
    last_err = None
    for k in range(retries):
        try:
            np.save(tmp, arr)
            if not tmp.exists():
                doubled = tmp.with_suffix(".npy.npy")
                if doubled.exists():
                    doubled.replace(tmp)
                else:
                    raise FileNotFoundError(
                        f"np.save did not produce expected file: {tmp}"
                    )
            _os.replace(tmp, path)
            return
        except OSError as e:
            last_err = e
            try:
                if tmp.exists():
                    tmp.unlink()
            except OSError:
                pass
            if e.errno not in (_errno.ETIMEDOUT, _errno.EAGAIN, _errno.EBUSY):
                raise
            _time.sleep(backoff ** k)
    raise last_err


# ─────────────────────────────────────────────────────────────────────────────
# Experiment runners
# ─────────────────────────────────────────────────────────────────────────────

def run_single_experiment(
    dataset_key: str,
    dataset_cfg: dict,
    df: pd.DataFrame,
    fp_text: str,
    model_cfg: dict,
    strategy: dict,
    run_idx: int,
    save_plot: bool = False,
) -> dict:
    """One repetition of one (dataset, model, strategy) cell."""
    features  = dataset_cfg["features"]
    gt_edges  = dataset_cfg["ground_truth_edges"]
    base      = make_base_prompt(
        dataset_cfg["label"], features, fp_text,
        dataset_cfg["feature_descriptions"],
    )
    prompt        = strategy["template"](base)
    n_internal    = int(strategy.get("self_consistency_runs", 1))

    raw_responses, latencies, errors = [], [], []
    for _ in range(n_internal):
        out = call_ollama(prompt, model=model_cfg["name"])
        raw_responses.append(out["response"])
        latencies.append(out["latency_s"])
        if not out["ok"]:
            errors.append(out["error"])

    if n_internal > 1:
        edge_sets   = [tuple(sorted(parse_causal_edges(r, features)))
                       for r in raw_responses]
        most_common = Counter(edge_sets).most_common(1)[0][0]
        edges       = list(most_common)
        used_text   = "\n---\n".join(raw_responses)
    else:
        edges     = parse_causal_edges(raw_responses[0], features)
        used_text = raw_responses[0]

    adj    = edges_to_adjacency(edges, features)
    scores = compute_evaluation_metrics_from_edges(edges, gt_edges, features)

    model_safe    = safe_name(model_cfg["name"])
    strategy_safe = safe_name(strategy["name"])
    out_dir       = REPEATED_RUNS_DIR / dataset_key / model_safe
    out_dir.mkdir(parents=True, exist_ok=True)
    run_tag       = f"{strategy_safe}__run{run_idx:02d}"

    _safe_save_npy(out_dir / f"{run_tag}.npy", adj.values.astype(np.int8))
    _safe_write_json(out_dir / f"{run_tag}.json", {
        "dataset":          dataset_key,
        "model":            model_cfg["name"],
        "strategy":         strategy["name"],
        "kind":             strategy["kind"],
        "run_idx":          run_idx,
        "features":         features,
        "edges":            edges,
        "scores":           scores,
        "latency_s_total":  round(sum(latencies), 3),
        "n_internal_calls": n_internal,
        "errors":           errors,
        "raw":              used_text,
    })

    if save_plot:
        plot_causal_graph(
            edges,
            title=f"{model_cfg['name']} | {strategy['name']} | run {run_idx} | {dataset_cfg['label']}",
            features=features,
            out_path=out_dir / f"{run_tag}.png",
            show=False,
        )

    return {
        "run_id":        uuid.uuid4().hex[:10],
        "run_idx":       run_idx,
        "dataset":       dataset_key,
        "dataset_label": dataset_cfg["label"],
        "model":         model_cfg["name"],
        "model_family":  model_cfg["family"],
        "strategy":      strategy["name"],
        "strategy_kind": strategy["kind"],
        "n_pred_edges":  int(adj.values.sum()),
        "edges":         edges,
        "adjacency":     adj.values.astype(np.int8),
        "latency_s":     round(sum(latencies), 3),
        "failed":        len(errors) == n_internal,
        "n_errors":      len(errors),
        **scores,
    }


def _make_failed_row(k, dataset_key, dataset_cfg, model_cfg, strategy, n_feat):
    """Build a placeholder row for a run that failed or was skipped by
    the watchdog. n_failed bucket counts these; aggregation handles them."""
    return {
        "run_id":        uuid.uuid4().hex[:10],
        "run_idx":       k,
        "dataset":       dataset_key,
        "dataset_label": dataset_cfg["label"],
        "model":         model_cfg["name"],
        "model_family":  model_cfg["family"],
        "strategy":      strategy["name"],
        "strategy_kind": strategy["kind"],
        "n_pred_edges":  0,
        "edges":         [],
        "adjacency":     np.zeros((n_feat, n_feat), dtype=np.int8),
        "latency_s":     0.0,
        "failed":        True,
        "n_errors":      1,
        "Precision":     0.0, "Recall": 0.0, "F1 Score": 0.0,
        "SHD":           0,   "FDR":    0.0,
        "Pearson Correlation": 0.0,
        "Hamming Distance":    0.0, "Frobenius Norm": 0.0,
    }


def run_repeated_experiments(
    dataset_key: str,
    dataset_cfg: dict,
    df: pd.DataFrame,
    fp_text: str,
    model_cfg: dict,
    strategy: dict,
    n_runs: int = N_RUNS,
    skip_existing: bool = False,
) -> list[dict]:
    """Run a single cell `n_runs` times.

    v3.7 changes:
      - per-cell wall-clock watchdog (CELL_BUDGET_S). Once exceeded, the
        remaining runs are recorded as failed and the loop returns. The
        next notebook run will re-attempt them via SKIP_EXISTING_RUNS.
    """
    rows = []
    n_feat = len(dataset_cfg["features"])
    model_safe    = safe_name(model_cfg["name"])
    strategy_safe = safe_name(strategy["name"])
    cell_t0 = _time.perf_counter()

    for k in range(1, n_runs + 1):
        # ── Watchdog: bail out if this cell has already used its budget ───
        elapsed = _time.perf_counter() - cell_t0
        if elapsed > CELL_BUDGET_S:
            print(f"         cell budget exceeded ({elapsed:.0f}s > {CELL_BUDGET_S}s) — "
                  f"skipping remaining reps {k}..{n_runs}")
            for kk in range(k, n_runs + 1):
                rows.append(_make_failed_row(kk, dataset_key, dataset_cfg,
                                             model_cfg, strategy, n_feat))
            break

        np.random.seed(RANDOM_SEED + k)

        run_tag   = f"{strategy_safe}__run{k:02d}"
        json_path = REPEATED_RUNS_DIR / dataset_key / model_safe / f"{run_tag}.json"
        adj_path  = REPEATED_RUNS_DIR / dataset_key / model_safe / f"{run_tag}.npy"

        # Resumability: if both artifacts exist and read cleanly, replay them.
        if skip_existing and json_path.exists() and adj_path.exists():
            try:
                cached = json.loads(json_path.read_text())
                adj    = np.load(adj_path)
            except Exception as e:
                print(f"      run {k:02d} cache read failed ({e!r}) — re-running")
            else:
                rows.append({
                    "run_id":        uuid.uuid4().hex[:10],
                    "run_idx":       k,
                    "dataset":       dataset_key,
                    "dataset_label": dataset_cfg["label"],
                    "model":         model_cfg["name"],
                    "model_family":  model_cfg["family"],
                    "strategy":      strategy["name"],
                    "strategy_kind": strategy["kind"],
                    "n_pred_edges":  int(adj.sum()),
                    "edges":         [tuple(e) for e in cached.get("edges", [])],
                    "adjacency":     adj,
                    "latency_s":     cached.get("latency_s_total", 0.0),
                    "failed":        len(cached.get("errors", [])) > 0,
                    "n_errors":      len(cached.get("errors", [])),
                    **cached["scores"],
                })
                continue

        try:
            row = run_single_experiment(
                dataset_key, dataset_cfg, df, fp_text,
                model_cfg, strategy, run_idx=k,
                save_plot=SAVE_PER_RUN_PLOTS,
            )
            rows.append(row)
        except Exception as e:
            print(f"      run {k:02d} FAILED: {type(e).__name__}: {e}")
            rows.append(_make_failed_row(k, dataset_key, dataset_cfg,
                                         model_cfg, strategy, n_feat))
    return rows


# ─────────────────────────────────────────────────────────────────────────────
# Aggregation (unchanged from v3.6)
# ─────────────────────────────────────────────────────────────────────────────

def aggregate_results(repeated_rows: list[dict],
                      features: list[str]) -> dict:
    """Aggregate `n_runs` rows for a single (dataset, model, strategy) cell."""
    n_runs = len(repeated_rows)
    n_feat = len(features)

    A = np.stack([r["adjacency"] for r in repeated_rows], axis=0)
    edge_prob = A.mean(axis=0)

    mean_metrics, std_metrics = {}, {}
    for m in EVAL_METRICS_ORDER:
        vals = np.array([r[m] for r in repeated_rows], dtype=float)
        mean_metrics[m] = float(np.mean(vals))
        std_metrics[m]  = float(np.std(vals, ddof=1) if n_runs > 1 else 0.0)

    n_failed  = int(sum(1 for r in repeated_rows if r["failed"]))
    n_zero    = int(sum(1 for r in repeated_rows
                        if (not r["failed"]) and r["n_pred_edges"] == 0))
    n_success = n_runs - n_failed - n_zero

    cell_var          = A.var(axis=0)
    consistency_pct   = 100.0 * (1.0 - cell_var.mean() / 0.25)
    consistency_pct   = float(max(0.0, min(100.0, consistency_pct)))

    return {
        "mean_metrics":     mean_metrics,
        "std_metrics":      std_metrics,
        "edge_prob_matrix": edge_prob,
        "n_runs":           n_runs,
        "n_success":        n_success,
        "n_zero":           n_zero,
        "n_failed":         n_failed,
        "consistency_pct":  consistency_pct,
    }


# ─────────────────────────────────────────────────────────────────────────────
# Inline smoke test (unchanged from v3.6)
# ─────────────────────────────────────────────────────────────────────────────
def _smoke_test_io():
    import tempfile
    with tempfile.TemporaryDirectory() as _td:
        _t = Path(_td)
        _safe_write_json(_t / "smoke.json", {"ok": True, "n": 1})
        assert (_t / "smoke.json").exists()
        _safe_save_npy(_t / "smoke.npy", np.array([[1, 0], [0, 1]], dtype=np.int8))
        assert (_t / "smoke.npy").exists()
        assert not (_t / "smoke.npy.npy").exists()
        loaded = np.load(_t / "smoke.npy")
        assert loaded.shape == (2, 2)
    return True

assert _smoke_test_io()
print("v3.7 modular functions defined: run_single_experiment, "
      "run_repeated_experiments (with watchdog), aggregate_results.")
print(f"CELL_BUDGET_S = {CELL_BUDGET_S}s ({CELL_BUDGET_S//60} min per cell)")
print("I/O smoke test: OK")
`````

`````python
# ─── §11.3 — Main repeated-run loop ─────────────────────────────────────────
# Resumable: per-run JSON sidecars are skipped on re-run. Safe to Ctrl+C and
# restart. Existing §7 single-run artifacts are NOT touched.
#
# To smoke-test the pipeline first, override before running:
#   MODELS_TO_RUN   = MODELS[:1]            # e.g. just llama3.1:8b
#   DATASETS_TO_RUN = ["tesla_ev"]
#   N_RUNS          = 3

raw_run_rows:    list[dict] = []   # one row per individual run
aggregated_rows: list[dict] = []   # one row per (dataset, model, strategy) cell

t_total = time.perf_counter()
total   = len(DATASETS_TO_RUN) * len(MODELS_TO_RUN) * len(STRATEGIES)
done    = 0

for ds_key in DATASETS_TO_RUN:
    ds_cfg   = DATASETS[ds_key]
    df_, fp_ = dataset_cache[ds_key]

    for m_cfg in MODELS_TO_RUN:
        for strat in STRATEGIES:
            done += 1
            label = f"{ds_key:<16} {m_cfg['name']:<22} {strat['name']}"
            print(f"[{done:>4}/{total}] {label}", flush=True)

            try:
                run_rows = run_repeated_experiments(
                    ds_key, ds_cfg, df_, fp_, m_cfg, strat,
                    n_runs=N_RUNS,
                    skip_existing=SKIP_EXISTING_RUNS,
                )
            except Exception as e:
                # Backstop only — run_repeated_experiments catches per-run
                # exceptions internally. Reaching this means the iterator
                # itself failed (e.g. a programming error).
                print(f"      CELL ABORTED: {type(e).__name__}: {e}")
                continue

            # Record raw (per-run) rows. Drop the in-memory adjacency before
            # serialisation — the .npy on disk is the persistent copy.
            for r in run_rows:
                rr = {k: v for k, v in r.items() if k != "adjacency"}
                rr["edges"] = [list(e) for e in rr["edges"]]
                raw_run_rows.append(rr)

            # Aggregate across the N_RUNS for this cell.
            agg = aggregate_results(run_rows, ds_cfg["features"])
            aggregated_rows.append({
                "dataset":         ds_key,
                "dataset_label":   ds_cfg["label"],
                "model":           m_cfg["name"],
                "model_family":    m_cfg["family"],
                "model_size_gb":   m_cfg["size_gb"],
                "strategy":        strat["name"],
                "strategy_kind":   strat["kind"],
                "row_label":       f"{m_cfg['name']}_{strat['name']}",
                **{f"{k}_mean": v for k, v in agg["mean_metrics"].items()},
                **{f"{k}_std":  v for k, v in agg["std_metrics"].items()},
                "n_runs":          agg["n_runs"],
                "n_success":       agg["n_success"],
                "n_zero":          agg["n_zero"],
                "n_failed":        agg["n_failed"],
                "consistency_pct": round(agg["consistency_pct"], 2),
            })

            # Persist the edge-probability matrix for this cell (atomic).
            cell_dir = REPEATED_RUNS_DIR / ds_key / safe_name(m_cfg["name"])
            cell_dir.mkdir(parents=True, exist_ok=True)
            try:
                _safe_save_npy(
                    cell_dir / f"{safe_name(strat['name'])}__edge_prob.npy",
                    agg["edge_prob_matrix"].astype(np.float32),
                )
            except OSError as e:
                print(f"      edge_prob save failed: {e!r}  (continuing)")

            # Console feedback for this cell.
            f1m = agg["mean_metrics"]["F1 Score"]
            f1s = agg["std_metrics"]["F1 Score"]
            print(f"      F1 = {f1m:.3f} ± {f1s:.3f}   "
                  f"n_zero={agg['n_zero']}/{N_RUNS}   "
                  f"n_failed={agg['n_failed']}   "
                  f"consistency={agg['consistency_pct']:.1f}%")

print(f"\nTotal wall-clock: {(time.perf_counter() - t_total) / 60:.1f} min")
print(f"Raw run rows:    {len(raw_run_rows):,}  (= cells × N_RUNS)")
print(f"Aggregated rows: {len(aggregated_rows):,}  (= cells)")


# ─────────────────────────────────────────────────────────────────────────────
# Durable persistence: retry on transient errors, fall back to pickle if CSV
# fails permanently. In-memory results are never lost to filesystem flakiness.
# ─────────────────────────────────────────────────────────────────────────────
def _durable_to_csv(df: pd.DataFrame, target: Path,
                    retries: int = 4, backoff: float = 1.5) -> None:
    last_err = None
    for k in range(retries):
        try:
            tmp = target.with_name(f"{target.name}.tmp-{_os.getpid()}")
            df.to_csv(tmp, index=False)
            _os.replace(tmp, target)
            print(f"  wrote {target}  ({len(df):,} rows)")
            return
        except Exception as e:
            last_err = e
            print(f"  attempt {k+1}/{retries} failed: {type(e).__name__}: {e}")
            _time.sleep(backoff ** k)
    # Last resort: pickle. Keeps the data, format is recoverable later.
    pkl = target.with_suffix(".pkl")
    df.to_pickle(pkl)
    print(f"     CSV failed permanently; saved as pickle -> {pkl}")
    print(f"     Recover with:  pd.read_pickle({str(pkl)!r}).to_csv({str(target)!r}, index=False)")

raw_runs_df   = pd.DataFrame(raw_run_rows)
aggregated_df = pd.DataFrame(aggregated_rows)

_durable_to_csv(raw_runs_df,   REPEATED_RUNS_DIR / "raw_runs.csv")
_durable_to_csv(aggregated_df, REPEATED_RUNS_DIR / "aggregated_results.csv")


# ─────────────────────────────────────────────────────────────────────────────
# Optional cloud sync (only if §11.1 detected a cloud-synced configured path).
# rsync is run once at the end, when nothing else competes for the volume.
# ─────────────────────────────────────────────────────────────────────────────
if SYNC_TARGET_DIR is not None:
    print(f"\nSyncing local cache to cloud: {SYNC_TARGET_DIR}")
    try:
        SYNC_TARGET_DIR.mkdir(parents=True, exist_ok=True)
    except OSError as e:
        print(f"   Could not create cloud target dir: {e!r}")
        print(f"   Local cache at {REPEATED_RUNS_DIR} is intact.")
    else:
        # shlex.quote handles spaces, '#', etc. in the OneDrive path safely.
        import shlex as _shlex
        _src = _shlex.quote(str(REPEATED_RUNS_DIR) + "/")
        _dst = _shlex.quote(str(SYNC_TARGET_DIR)   + "/")
        rc = _os.system(f"rsync -a --info=progress2 {_src} {_dst}")
        if rc == 0:
            print("Sync OK.")
        else:
            print(f"   rsync exit code {rc}.")
            print(f"   Local cache at {REPEATED_RUNS_DIR} still has everything.")
            print(f"   Re-run the sync manually when convenient:")
            print(f"     rsync -a {_src} {_dst}")
`````

## 12. Aggregated Average Benchmark Matrix

Headline result table: **rows = `(model, strategy)` cells, columns = the
8 evaluation metrics, values = mean across `N_RUNS` runs**. A companion
std matrix (same shape, values = sample standard deviation) is built
alongside, and a *detailed statistics* table records mean / std /
n_success / n_zero / n_failed / consistency for every cell.

The final cell of this section prints per-model and per-strategy
summaries so the analysis section of a paper has tabular numbers ready
to quote without re-pivoting in pandas.

`````python
# ─── §12.1 — Build the average and std benchmark matrices ───────────────────
def build_average_benchmark_matrix(
    aggregated_rows: list[dict],
    dataset_key: str,
    metric_kind: str = "mean",
) -> pd.DataFrame:
    """One row per (model, strategy) cell, columns = the 8 metrics.

    Parameters
    ----------
    metric_kind : "mean" | "std"
        "mean" -> mean of metric across runs (the average benchmark matrix)
        "std"  -> sample std across runs       (the variability matrix)
    """
    rows = [r for r in aggregated_rows if r["dataset"] == dataset_key]
    if not rows:
        return pd.DataFrame()
    suffix = "_mean" if metric_kind == "mean" else "_std"
    cols   = [f"{m}{suffix}" for m in EVAL_METRICS_ORDER]
    df     = pd.DataFrame(rows).set_index("row_label")[cols]
    df.columns = EVAL_METRICS_ORDER
    return df


average_matrices: dict[str, pd.DataFrame] = {}
std_matrices:     dict[str, pd.DataFrame] = {}

for ds_key in DATASETS_TO_RUN:
    avg = build_average_benchmark_matrix(aggregated_rows, ds_key, "mean")
    sd  = build_average_benchmark_matrix(aggregated_rows, ds_key, "std")
    average_matrices[ds_key] = avg
    std_matrices[ds_key]     = sd
    if not avg.empty:
        _durable_to_csv(avg, REPEATED_RUNS_DIR / f"average_benchmark_matrix_{ds_key}.csv")
        _durable_to_csv(sd,  REPEATED_RUNS_DIR / f"std_benchmark_matrix_{ds_key}.csv")

    print(f"\n=== Average Benchmark Matrix — "
          f"{DATASETS[ds_key]['label']}  (N_RUNS={N_RUNS}) ===")
    print(f"Shape: {avg.shape}")
    if not avg.empty:
        print(avg.round(3).head(15).to_string())
`````

`````python
# ─── §12.2 — Detailed statistics table ──────────────────────────────────────
# One row per (dataset, model, strategy) cell, with mean+std for every metric
# and the success / zero-detection / failure counts that motivated the
# repeated-run section in the first place.

stat_cols = (
    ["dataset", "model", "strategy", "strategy_kind",
     "n_runs", "n_success", "n_zero", "n_failed", "consistency_pct"]
    + [f"{m}_mean" for m in EVAL_METRICS_ORDER]
    + [f"{m}_std"  for m in EVAL_METRICS_ORDER]
)
detailed_stats_df = pd.DataFrame(aggregated_rows)[stat_cols].copy()

# Round float columns for human-readable display before persisting.
float_cols = [c for c in detailed_stats_df.columns
              if detailed_stats_df[c].dtype.kind == "f"]
detailed_stats_df[float_cols] = detailed_stats_df[float_cols].round(4)

_durable_to_csv(detailed_stats_df, REPEATED_RUNS_DIR / "detailed_statistics.csv")

# Quick skim view: F1 mean/std + the counts.
display_cols = [
    "dataset", "model", "strategy",
    "F1 Score_mean", "F1 Score_std",
    "n_success", "n_zero", "n_failed", "consistency_pct",
]
print("\nDetailed statistics (head):")
print(detailed_stats_df[display_cols].head(20).to_string(index=False))


# ── Per-model and per-strategy summaries ───────────────────────────────────
for ds_key in DATASETS_TO_RUN:
    sub = detailed_stats_df[detailed_stats_df["dataset"] == ds_key]
    if sub.empty:
        continue
    label = DATASETS[ds_key]["label"]

    print(f"\n=== Per-model F1 (mean ± std)  —  {label} ===")
    pm = (sub.groupby("model")[["F1 Score_mean", "F1 Score_std",
                                 "consistency_pct", "n_zero"]]
              .mean().round(3)
              .sort_values("F1 Score_mean", ascending=False))
    print(pm.to_string())

    print(f"\n=== Per-strategy F1 (mean ± std)  —  {label} ===")
    ps = (sub.groupby("strategy")[["F1 Score_mean", "F1 Score_std",
                                    "consistency_pct", "n_zero"]]
              .mean().round(3)
              .sort_values("F1 Score_mean", ascending=False))
    print(ps.to_string())
`````

## 13. Heatmap Visualisation of the Average Benchmark Matrix

Three heatmaps are produced per dataset:

* **13.1 — Full averaged matrix** *(rows: every `(model, strategy)` cell;
  columns: 8 metrics; cell text: `mean ± std`)*. Single `coolwarm`
  colormap, value annotations inside every cell, readable axis labels —
  drop-in for a thesis figure.
* **13.2a — Compact per-model heatmap** *(rows: 10 LLMs, columns: 8 metrics)*.
  Values averaged over all strategies within each model.
* **13.2b — Compact per-strategy heatmap** *(rows: 17+3 strategies,
  columns: 8 metrics)*. Values averaged over all models within each strategy.

`````python
# ─── §13.1 — Full averaged matrix heatmap (screenshot-style) ────────────────
def plot_average_matrix(
    avg_df: pd.DataFrame,
    std_df: pd.DataFrame | None,
    dataset_key: str,
    show_std: bool = True,
    cmap: str = "coolwarm",
):
    """Heatmap of the average benchmark matrix.

    rows  = (model, strategy) cell labels
    cols  = the 8 evaluation metrics
    cell  = annotated with `mean` (or `mean\\n±std` if show_std=True)

    Note: SHD, FDR, Hamming and Frobenius are *lower-is-better*; the colour
    mapping does not flip them. Read those columns with that in mind.
    """
    if avg_df.empty:
        print(f"[{dataset_key}] no data")
        return

    if show_std and std_df is not None and not std_df.empty:
        annot = (avg_df.round(3).astype(str)
                 + "\n±"
                 + std_df.reindex_like(avg_df).round(3).astype(str))
        annot_fmt = ""
    else:
        annot     = avg_df.round(4)
        annot_fmt = ".4f"

    fig_h = max(0.42 * len(avg_df) + 2, 6)
    fig_w = 2.0 * len(EVAL_METRICS_ORDER) + 4
    fig, ax = plt.subplots(figsize=(fig_w, fig_h))

    sns.heatmap(
        avg_df,
        ax=ax,
        annot=annot, fmt=annot_fmt,
        cmap=cmap,
        cbar=True,
        linewidths=0.4, linecolor="white",
        annot_kws={"fontsize": 7},
    )
    ax.set_title(
        f"Average Evaluation Metrics for Causal Discovery Models  "
        f"(N_RUNS = {N_RUNS}) — {DATASETS[dataset_key]['label']}",
        fontsize=12,
    )
    ax.set_xlabel("")
    ax.set_ylabel("Causal Discovery Models  (LLM × Prompting Strategy)",
                  fontsize=10)
    plt.xticks(rotation=30, ha="right", fontsize=9)
    plt.yticks(fontsize=7)
    plt.tight_layout()

    out = REPEATED_RUNS_DIR / f"average_benchmark_heatmap_{dataset_key}.png"
    try:
        plt.savefig(out, dpi=140, bbox_inches="tight")
        print(f"Saved: {out}")
    except OSError as e:
        print(f"   Could not save heatmap to {out}: {e!r}")
    plt.show()


for ds_key in DATASETS_TO_RUN:
    plot_average_matrix(
        average_matrices.get(ds_key, pd.DataFrame()),
        std_matrices.get(ds_key, pd.DataFrame()),
        ds_key,
        show_std=True,
    )
`````

`````python
# ─── §13.2 — Compact heatmaps: per-model and per-strategy ───────────────────
# Each view averages the metric across the orthogonal axis, so each heatmap
# has 10–20 rows — the same row count as the screenshot reference style.

def _compact_heatmap(
    pivot: pd.DataFrame,
    title: str,
    out_path: Path,
    cmap: str = "coolwarm",
):
    if pivot.empty:
        return
    fig_h = max(0.45 * len(pivot) + 2, 5)
    fig_w = 1.9 * pivot.shape[1] + 4
    fig, ax = plt.subplots(figsize=(fig_w, fig_h))
    sns.heatmap(
        pivot, ax=ax, annot=True, fmt=".3f",
        cmap=cmap, cbar=True,
        linewidths=0.4, linecolor="white",
        annot_kws={"fontsize": 9},
    )
    ax.set_title(title, fontsize=12)
    ax.set_xlabel(""); ax.set_ylabel("")
    plt.xticks(rotation=30, ha="right", fontsize=10)
    plt.yticks(fontsize=10)
    plt.tight_layout()
    try:
        plt.savefig(out_path, dpi=140, bbox_inches="tight")
        print(f"Saved: {out_path}")
    except OSError as e:
        print(f"   Could not save heatmap to {out_path}: {e!r}")
    plt.show()


for ds_key in DATASETS_TO_RUN:
    sub = detailed_stats_df[detailed_stats_df["dataset"] == ds_key]
    if sub.empty:
        continue
    label     = DATASETS[ds_key]["label"]
    mean_cols = [f"{m}_mean" for m in EVAL_METRICS_ORDER]

    # 13.2a — Per-model (rows = LLM, cols = 8 metrics, value = mean over strategies)
    per_model = (sub.groupby("model")[mean_cols].mean().round(4))
    per_model.columns = EVAL_METRICS_ORDER
    _durable_to_csv(per_model.reset_index(),
                    REPEATED_RUNS_DIR / f"per_model_avg_{ds_key}.csv")
    _compact_heatmap(
        per_model,
        title=f"Per-Model Average Evaluation (N_RUNS = {N_RUNS}) — {label}",
        out_path=REPEATED_RUNS_DIR / f"per_model_avg_heatmap_{ds_key}.png",
    )

    # 13.2b — Per-strategy (rows = strategy, cols = 8 metrics, value = mean over models)
    strat_order = [s["name"] for s in STRATEGIES if s["name"] in sub["strategy"].unique()]
    per_strat = (sub.groupby("strategy")[mean_cols].mean()
                    .reindex(strat_order).round(4))
    per_strat.columns = EVAL_METRICS_ORDER
    _durable_to_csv(per_strat.reset_index(),
                    REPEATED_RUNS_DIR / f"per_strategy_avg_{ds_key}.csv")
    _compact_heatmap(
        per_strat,
        title=f"Per-Strategy Average Evaluation (N_RUNS = {N_RUNS}) — {label}",
        out_path=REPEATED_RUNS_DIR / f"per_strategy_avg_heatmap_{ds_key}.png",
    )
`````
